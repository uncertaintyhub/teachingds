## The Lotka-Volterra model (Task 1)

... is a pair of first-order, non-linear, differential equations used to describe the dynamics of biological systems in which two species interact, one as a predator and the other as prey. The explicit Euler scheme for the computational model is given by:

$$
\mathbf{z}_{n+1} = \Psi(\mathbf{z}_{n}) = \begin{pmatrix}R_n + \Delta t \left( \alpha R_n - \beta R_n F_n \right)\\
F_n + \Delta t \left( \delta R_n F_n - \gamma F_n \right) \end{pmatrix}
$$
where:
- $\mathbf{z}_n=(R_n,F_n)^\top$ denote the population state at discrete time step $n$ (where $t_n = n \Delta t$)
- $ R_n $ is the population of rabbits at time step $ n $.
- $ F_n $ is the population of foxes at time step $ n $.
- $ \Delta t $ is the time step size.
- $ \alpha $ is the natural growth rate of prey.
- $ \beta $ is the natural death rate of prey due to predation.
- $ \gamma $ is the natural death rate of predators.
- $ \delta $ is the natural growth rate of predators due to predation.
